Q8.1
NameError: name 'val' is not defined
Explanation: val is a local variable that exists only inside the function. It cannot be accessed outside the function.

```python
#Q8.2
counter = 0
def show_counter():
    print(counter)
show_counter()
```

```python
#Q8.3
counter = 0
def increment_counter():
    counter = counter + 1
increment_counter()
```

```python
#Q8.4
counter = 0
def increment_counter():
    global counter
    counter = counter + 1
increment_counter()
print(counter)
```
