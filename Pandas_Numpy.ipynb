```python
#Q36
a = np.arange(21, 46).reshape(5, 5)
central = a[1:4, 1:4]
transposed = central.T
product = central @ transposed
row_sums = product.sum(axis=1)
maximum = row_sums.max()
positions = np.where(row_sums == maximum)[0]
print(a)
print(central)
print(transposed)
print(product)
print(row_sums)
print(maximum)
print(positions)
```

```python
#Q37
quantity = np.array([[20, 15, 30, 10],
    [25, 18, 22, 12],
    [30, 20, 25, 15],
    [22, 17, 28, 11],
    [28, 19, 20, 14],
    [24, 16, 26, 13]])
original_prices = np.array([150, 240, 90, 320])
daily_revenue = quantity @ original_prices
product_quantity = quantity.sum(axis=0)
product_revenue = product_quantity * original_prices
best_day = daily_revenue.argmax() + 1
best_product = product_quantity.argmax() + 1
discounted_prices = np.where(
    product_quantity > 100,
    original_prices * 0.90,
    original_prices)
new_total_revenue = quantity.sum(axis=0) @ discounted_prices
print(daily_revenue)
print(product_revenue)
print(daily_revenue.sum())
print(best_day)
print(product_quantity)
print(best_product)
print(discounted_prices)
print(new_total_revenue)
print(original_prices)
```

```python
#Q38
np.random.seed(42)
data = np.random.normal(50, 12, 100)
outside = (data < 30) | (data > 70)
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))
print(outside.sum())
print(data.mean(), cleaned.mean())
print(data.var(), cleaned.var())
print(data.std(), cleaned.std())
print(data.min(), cleaned.min())
print(data.max(), cleaned.max())
```

```python
#Q39
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})
ordinary = branch_a + branch_b + branch_c
print(ordinary)
common = branch_a.index.intersection(branch_b.index)
common = common.intersection(branch_c.index)
print(common)
combined_common = (branch_a[common] + branch_b[common] + branch_c[common])
print(combined_common)
zero_result = (branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0))
print(zero_result)
```

```python
#Q40
scores = {"R101": 78,"R102": 65,"R104": 92,"R105": 48}
required = ["R105", "R103", "R101", "R106", "R104", "R102"]
s = pd.Series(scores, index=required)
print(s)
print(s[s.isnull()])
valid = s[s.notnull() & (s >= 70)]
print(valid)
data = valid.to_dict()
print(data)
new_series = pd.Series(data)
print(new_series)
print(s[valid.index].equals(new_series))
```
